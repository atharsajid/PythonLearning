Question 1
Student vs Money
Registration fee = 500
Student fee = 2000
What will be the plotting graph if 4 students take a registration?

```python
import matplotlib.pyplot as plt

registration_fees = 500
student_fees = 2000
no_of_students = 400

y = []
x = []
for i in range(no_of_students+1):
    x.append(i)
    value = (student_fees * i) + registration_fees
    y.append(value)

plt.plot(x, y)
plt.xlabel("No of Students")
plt.ylabel("Fees")
plt.draw()
```

1. Plot the graph of:

y=2x+1
Create a table for x=−2,−1,0,1,2.
Plot the points on a graph.
Draw the straight line.

```python
import matplotlib.pyplot as plt

x = [-2, -1, 0, 1, 2]
y = []

for i in x:
    value = (2*i) + 1
    y.append(value)

plt.plot(x, y)
plt.xlabel("X")
plt.ylabel("Y")

plt.draw()


```

2. Plot the graph of:

y=−x+3
Find at least 5 points.
Identify where the line intersects the x-axis and y-axis.

```python
import matplotlib.pyplot as plt


# y=−x+3

no_of_points = 3
y = []
x = []

for i in range(-(no_of_points), no_of_points+1):
    x.append(i)
    value = -(i) + 3
    y.append(value)


plt.plot(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.show()

```

```python
import matplotlib.pyplot as plt

study_time = [2,4,6,8,10]
exam_score = [60,70,80,85,90]
exam_score_expected = [75,80,86,90,95]

plt.scatter(study_time, exam_score, marker="^",color="green", label = "Actual Score")
plt.scatter(study_time, exam_score_expected, marker="^",color="orange", label = "Expected Score")
plt.legend()
plt.xlabel("Study Hour")
plt.ylabel("Exam Score")
plt.title("Relationship between Study Time and Exam Score")
plt.grid(linestyle = "-.")

plt.show()
```
